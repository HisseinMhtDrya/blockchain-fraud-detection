```python

from pathlib import Path
import pandas as pd

DATA_DIR = Path(
    r"C:\Users\hisse\blockchain-fraud-detection\data\raw\archive\elliptic_bitcoin_dataset"
)

# Charger les trois fichiers
classes = pd.read_csv(DATA_DIR / "elliptic_txs_classes.csv")
edges = pd.read_csv(DATA_DIR / "elliptic_txs_edgelist.csv", nrows=5)
features = pd.read_csv(
    DATA_DIR / "elliptic_txs_features.csv",
    header=None,
    nrows=5
)

print("=== CLASSES DES TRANSACTIONS ===")
print(classes.head())
print("\nDimensions du fichier classes :", classes.shape)

print("\n=== LIENS ENTRE TRANSACTIONS ===")
print(edges.head())
print("\nColonnes :", edges.columns.tolist())

print("\n=== CARACTÉRISTIQUES DES TRANSACTIONS ===")
print(features.iloc[:, :6])
print("\nNombre de colonnes du fichier features :", features.shape[1])
```

```python

# Afficher la répartition des classes
print("Répartition des classes :")
print(classes["class"].value_counts())

print("\nPourcentage de chaque classe :")
print(
    (classes["class"].value_counts(normalize=True) * 100)
    .round(2)
    .astype(str) + "%"
)

print("\nValeurs uniques :")
print(classes["class"].unique())
```

```python

import numpy as np
import pandas as pd

# Préparer les étiquettes connues
labels = classes.copy()
labels["class"] = labels["class"].astype(str)

labels = labels[labels["class"].isin(["1", "2"])].copy()
labels["target"] = labels["class"].map({"1": 0, "2": 1})

# Charger les caractéristiques
features = pd.read_csv(
    DATA_DIR / "elliptic_txs_features.csv",
    header=None
)

# La première colonne est l'identifiant de transaction.
# La deuxième colonne est l'indice temporel.
feature_names = (
    ["time_step"] +
    [f"feature_{i}" for i in range(1, features.shape[1] - 1)]
)

features.columns = ["txId"] + feature_names

# Associer chaque transaction à ses caractéristiques et à sa classe
dataset = labels[["txId", "target"]].merge(
    features,
    on="txId",
    how="inner",
    validate="one_to_one"
)

print("Dimensions du jeu de données :", dataset.shape)
print("\nRépartition des cibles :")
print(dataset["target"].value_counts().rename(index={
    0: "Licite",
    1: "Illicite"
}))
print("\nValeurs manquantes :", dataset.isna().sum().sum())
print("\nAperçu :")
display(dataset.head())
```

```python

# Séparation chronologique : périodes 1 à 34 pour l'entraînement,
# périodes 35 à 49 pour le test.

train_df = dataset[dataset["time_step"] <= 34].copy()
test_df = dataset[dataset["time_step"] > 34].copy()

# Ne pas utiliser l'identifiant comme caractéristique du modèle
feature_cols = [
    col for col in dataset.columns
    if col not in ["txId", "target"]
]

X_train = train_df[feature_cols]
y_train = train_df["target"]

X_test = test_df[feature_cols]
y_test = test_df["target"]

print("=== DIMENSIONS ===")
print("Entraînement :", X_train.shape)
print("Test :", X_test.shape)

print("\n=== CLASSES D'ENTRAÎNEMENT ===")
print(y_train.value_counts().rename(index={
    0: "Licite",
    1: "Illicite"
}))

print("\n=== CLASSES DE TEST ===")
print(y_test.value_counts().rename(index={
    0: "Licite",
    1: "Illicite"
}))

print("\nPériodes d'entraînement :",
      train_df["time_step"].min(), "à",
      train_df["time_step"].max())

print("Périodes de test :",
      test_df["time_step"].min(), "à",
      test_df["time_step"].max())
```
